[Operator overloading](https://realpython.com/operator-function-overloading/) is a useful concept of Python classes that you may find useful when defining your own objects.

In the example below, I overload the `&` operator of my `Ket` object to represent the tensor product (done by `np.kron`) of the kets on the left and right of the operator.

```python
class Ket:
    def __init__(self, state=[0, 0], n=1):
        self.state = np.array(state, dtype=np.complex128)
        self.n = n

    def __repr__(self):
        return f'{self.state}'
    
    def __and__(self, ket2):
        if isinstance(ket2, Ket):
            state = np.kron(self.state, ket2.state)
            n = self.n + ket2.n
            return Ket(state, n)

```

```python
psi = Ket([1, 0])
phi = Ket(1/np.sqrt(2) * np.array([1, 1]))
```

```python
psi
```

```python
xi = psi & phi
xi.n, xi.state
```

```python
zeta = phi & xi
zeta.n, zeta.state
```
