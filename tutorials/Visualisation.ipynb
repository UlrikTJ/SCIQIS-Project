# Visualisation

Let's look at three different ways of illustrating the basics of [amplitude modulation (AM)](https://en.wikipedia.org/wiki/Amplitude_modulation) and [phase modulation (PM)](https://en.wikipedia.org/wiki/Phase_modulation) (which is related to frequency modulation (FM)).

A transmitted analog signal with a fixed _carrier frequency_ $f_c$ and with variable amplitude $A(t)$ and phase $\phi(t)$  has the waveform 
$$
s(t) = A(t) \cos(2\pi f_c t + \phi(t))
$$

For AM, the signal to be transmitted is encoded into the amplitude $A$ while the phase $\phi$ is kept constant. Conversely for PM.

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact
```

```python
def s(t, fc, A, phi):
    return A * np.cos(2*np.pi * fc * t + phi)

def sig2AM(signal, depth):
    """
    signal : array-like
        Input signal to be encoded
    depth : float
        Modulation depth, between 0 and 1
    """
    signal = np.array(signal)
    signal_normalised = signal # / signal.max()  # remove this, as it interferes with animation further down
    modulation = 1 + depth * signal_normalised
    return modulation

def sig2PM(signal, depth):
    """
    signal : array-like
        Input signal to be encoded
    depth : float
        Modulation depth, between 0 and 1 (will be scaled to pi)
    """
    signal = np.array(signal)
    signal_normalised = signal # / signal.max()  # remove this, as it interferes with animation further down
    modulation = 1 + np.pi * depth * signal_normalised
    return modulation

def AM(t, signal, fc=10, depth=.5):
    modulation = sig2AM(signal, depth)
    encoded = s(t, fc, modulation, 0)
    return encoded

def PM(t, signal, fc=10, depth=.5):
    modulation = sig2PM(signal, depth)
    encoded = s(t, fc, 1, modulation)
    return encoded
```

```python
t_max = 2.
dt = .001
t = np.arange(0, t_max, dt)

signal = (t - t_max/2) * np.exp(-(t - t_max/2)**2 / (2 * .2**2))  # some arbitrary, interesting-looking signal
signal = np.cos(2*np.pi * 2 * t)

plt.plot(t, signal)
```

You can get more interesting interactions with the [ipywidgets](https://ipywidgets.readthedocs.io/en/latest/index.html) library that enables interactive data exploration, dashboards, etc. It can in principle even be used to create simple GUI applications. The easiest way to get started is to use the simple [interact](https://ipywidgets.readthedocs.io/en/latest/examples/Using%20Interact.html) interface:

```python
t_max = 2.
dt = .001
t = np.arange(0, t_max, dt)

@interact(fc=(1.,20.), f_signal=(0., 10.), depth=(0., 1.), show_signal=True)
def plot_modsig(fc=10, f_signal=2, depth=.5, show_signal=True):
    signal = np.cos(2*np.pi * f_signal * t)
    
    fig, (ax0, ax1) = plt.subplots(2, 1, figsize=(8,5), sharex=True,
                                   gridspec_kw=dict(hspace=0.05))  # see https://matplotlib.org/stable/users/explain/axes/arranging_axes.html
    ax0.plot(t, AM(t, signal, fc=fc, depth=depth))
    ax1.plot(t, PM(t, signal, fc=fc, depth=depth))
    
    ax0.set_ylabel('AM signal')
    ax1.set_ylabel('PM signal')
    ax1.set_xlabel('time [s]')
    
    for ax in [ax0, ax1]:
        if show_signal:
            ax.plot(t, signal, alpha=.6)
        
        ax.set_xlim(0, t_max)
        ax.set_ylim(-2.1, 2.1)
        
        ax.set_facecolor('floralwhite')  # see https://matplotlib.org/stable/gallery/color/named_colors.html
    
        ax.grid(True)
        ax.spines[:].set_visible(False)
        ax.tick_params(bottom=False, left=False)

    plt.show()
```

## Animation

Matplotlib makes it relatively easy to create [simple animations](https://matplotlib.org/stable/users/explain/animations/animations.html#sphx-glr-users-explain-animations-animations-py), particularly useful for plots, but in principle the many different Artists makes it possible to create rather advanced animations.

There are two different interfaces: `FuncAnimation`, which takes parametrised instructions, and `ArtistAnimation` which animates a list of pre-created figures.

```python
from matplotlib import animation
from IPython.display import HTML

plt.rcParams["font.family"] = "Roboto Slab"
plt.rcParams["font.size"] = 14

fig, (ax0, ax1) = plt.subplots(2, 1, figsize=(8,5), sharex=True,
                               gridspec_kw=dict(hspace=0.05))  # see https://matplotlib.org/stable/users/explain/axes/arranging_axes.html
line0, = ax0.plot(t, AM(t, signal, depth=1))
line1, = ax1.plot(t, PM(t, signal, depth=1))

ax0.set_ylabel('AM signal')
ax1.set_ylabel('PM signal')
ax1.set_xlabel('time [s]')


line0s, = ax0.plot(t, signal, alpha=.6)
line1s, = ax1.plot(t, signal, alpha=.6)

for ax in [ax0, ax1]:
    
    ax.set_xlim(0, t_max)
    ax.set_ylim(-2.1, 2.1)
    
    ax.set_facecolor('floralwhite')  # see https://matplotlib.org/stable/gallery/color/named_colors.html

    ax.grid(True)
    ax.spines[:].set_visible(False)
    ax.tick_params(bottom=False, left=False)

# specify animated parameters
a_signal = np.ones(180)
a_signal[:30] = np.linspace(1, 0, 30)
a_signal[30:60] = np.linspace(0, 1, 30)
f_signal = 2 * np.ones(180)
f_signal[60:120] = np.linspace(2, 6, 60)
f_signal[120:180] = np.linspace(6, 2, 60)

def update(frame):
    signal = a_signal[frame] * np.cos(2*np.pi * f_signal[frame] * t)
    line0s.set_ydata(signal)
    line1s.set_ydata(signal)
    line0.set_ydata(AM(t, signal, depth=1))
    line1.set_ydata(PM(t, signal, depth=1))
    return (line0, line1, line0s, line1s)

anim = animation.FuncAnimation(fig=fig, func=update, frames=180, interval=40)
HTML(anim.to_html5_video());
```
