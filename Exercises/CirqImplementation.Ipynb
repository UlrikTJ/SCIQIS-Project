```python
import cirq
import sympy


angle_vals = np.linspace(0, 2 * np.pi, 1000)
p000_list = []

theta = sympy.Symbol('theta')

circuit = cirq.Circuit()
qubits = cirq.LineQubit.range(3)

circuit.append(cirq.ry(theta)(qubits[0]))
circuit.append(cirq.ry(theta)(qubits[1]))
circuit.append(cirq.ry(theta)(qubits[2]))

not_op = cirq.CNOT(qubits[0], qubits[1])
not_op2 = cirq.CNOT(qubits[1], qubits[2])

circuit.append(not_op)
circuit.append(not_op2)

circuit.append(cirq.ry(theta)(qubits[0]))
circuit.append(cirq.ry(theta)(qubits[1]))
circuit.append(cirq.ry(theta)(qubits[2]))

not_op = cirq.CNOT(qubits[0], qubits[1])
not_op2 = cirq.CNOT(qubits[1], qubits[2])

circuit.append(not_op)
circuit.append(not_op2)

circuit.append(cirq.measure(qubits[0], qubits[1],qubits[2], key='result'))

#print(circuit)
```

```python
s = cirq.Simulator()

print('Simulate the circuit:')
results = s.simulate(circuit)
print(results)


samples = s.run(circuit, repetitions=1000)

samples
```

```python
import matplotlib.pyplot as plt

cirq.plot_state_histogram(samples, plt.subplot())
plt.show()
```
