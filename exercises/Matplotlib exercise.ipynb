```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def model(t, a, b):
    # return 1 + x**a * np.sin(b*x)
    return np.cos(a*t) * np.exp(-b*t)
```

```python
rng = np.random.default_rng(42)

xsamples = np.linspace(0, 3, 17)
y1 = model(xsamples, 10, 1) + rng.normal(scale=.1, size=xsamples.shape)
y2 = model(xsamples, 20, 1.4) + rng.normal(scale=.1, size=xsamples.shape)
```

```python
x = np.linspace(0, 3, 200)
plt.plot(x, model(x, 10, 1))
plt.plot(x, model(x, 20, 1.4))

plt.scatter(xsamples, y1)
plt.scatter(xsamples, y2)
# plt.plot(x, model(x, 5, 2))
# plt.plot(x, model(x, 8, .4))
```
